# 01 Data exploration (plan)

Planned data-quality checks for FloodShield Phase 2. This notebook contains no
observations yet; code cells will read the files produced by
`scripts/select_gauges.py` and `scripts/build_forcing_table.py`
(CAMELS-IND streamflow and IMD / IMDAA forcings). Open-Meteo is not used.

## 1. Gauge availability per basin

- Godavari (CWC code `03`) and Mahanadi (`08`) gauges in the "streamflow sufficient" subset.
- Selected gauges and reasons: `configs/selected_gauges.csv`, `outputs/gauge_selection_report.csv`.

## 2. Missing-data patterns

- Fraction of non-missing daily streamflow per gauge, per year and per season.
- The shared Godavari gap around 2014-2015 and the early-2019 end of observations.

## 3. Flow distributions

- Per-gauge flow distributions (log scale), seasonal cycle, annual maxima.
- Training-period high-flow thresholds (`configs/flow_thresholds.csv`) and events per split.
- Gauges downstream of large dams where flows are regulated.

## 4. Units and forcing checks

- Streamflow in m3/s; forcing units from the file headers (`camels_forcings_selected_units.json`).
- Known forcing anomalies (e.g. tmin = 0.00 C at every gauge on 2019-01-11): `docs/live_data_options.md`.

## 5. Weather and streamflow date alignment

- IMD rainfall day definition versus the CAMELS-IND streamflow day (day boundary not stated in the documents).
- Verify there is no off-by-one-day shift between rain inputs and flow targets (e.g. lagged cross-correlation).
- Live: IMD real-time grids (rain 0.25°, temperature 0.5°) versus the archive (temperature 1°) used for training.
- Split periods: `configs/training.yaml` (train 1991-2008, validation 2009-2012, test 2013-2018).
